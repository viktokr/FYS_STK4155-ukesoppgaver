Using ridge through SVD instead of open form (XᵀX + λI)θ = Xᵀy means that X is decomposed with SVD first, creating a sum over singular values instead of solving XᵀX.
$\\$
XᵀX squares the condition number (how numerically sensitive the matrix X is to work with) in X, potentially creating really bad (high) condition value for XᵀX even if X is only moderately ill-conditioned. Furthermore, this often happens for polynomials of high degrees (x^15) which is seen in exercise 3.$\\$
Sidenote: Condition number - high condition number means small changes in your data can produce large changes in the computed solution θ. It's a measure of how numerically unstable solving a system involving X will be.
$\\$
SVD doesn't use XᵀX and is therefor a more numerically reliable method. 

How ridge works by a 'shrinking factor': σᵢ²/(σᵢ²+λ)

Ridge solution via SVD: θ = Σᵢ [σᵢ/(σᵢ²+λ)] (uᵢ·y) vᵢ. 
Compared to plain OLS via SVD, θ_OLS = Σᵢ [1/σᵢ] (uᵢ·y) vᵢ. 

The ridge version can be rewritten as:
θ_ridge = Σᵢ [σᵢ²/(σᵢ²+λ)] · [1/σᵢ (uᵢ·y)] · vᵢ = Σᵢ [σᵢ²/(σᵢ²+λ)] · θ_OLS,ᵢ · vᵢ

So ridge takes the OLS solution and multiplies each direction's contribution by the factor σᵢ²/(σᵢ²+λ), which is always between 0 and 1:

If σᵢ is large (well-determined direction, lots of signal): σᵢ²/(σᵢ²+λ) ≈ 1 → barely shrunk, ridge trusts this direction.
If σᵢ is small (noisy, poorly-determined direction): σᵢ²/(σᵢ²+λ) ≈ 0 → shrunk almost to zero, ridge distrusts this direction.

Bias is the error from your model being too simple to represent the true underlying pattern — systematically missing something regardless of how much data you have. A degree-1 line trying to fit your double-Gaussian bump data has high bias: it can't capture that shape no matter what training set it sees, because a straight line just isn't flexible enough.

Variance is the error from your model being too sensitive to the specific noisy training sample it happened to see. A degree-15 polynomial has enough flexibility to snake through every little noise wiggle in your particular 67 training points — but if you resampled a different random training set (different noise draws), you'd get a wildly different degree-15 curve. That instability — the model changing a lot depending on which specific noisy sample it was trained on — is variance.


These two move in opposite directions as you increase model flexibility (polynomial degree, or decreasing λ in Ridge):

Low complexity (degree 1, or huge λ) → high bias, low variance. The model is stable (barely reacts to which training points it sees) but systematically wrong/underfit.

High complexity (degree 15, or tiny λ) → low bias, high variance. The model can represent the true shape, but it also greedily fits noise, so it swings wildly depending on the specific training sample — overfit.

And it's exactly what Ridge's λ is doing too, just via a different dial: 
large λ → heavy shrinkage → low variance but you're forcing coefficients toward 0 even in directions where there's real signal → some bias reintroduced. 
Small λ → the fit is closer to unregularized OLS on 15 features → low bias but high variance again. 

The dfs (effective degrees of freedom) is precisely tracking where you sit on this same bias-variance axis — 
high df ≈ low bias/high variance end, 
low df ≈ high bias/low variance end 
— just continuously instead of in the discrete jumps that changing polynomial degree gives you.

We use the Hessian matrix in gradient descent because it describes how the slope (gradient) of the (cost)-function changes in different directions. It helps confirm if a point is a local minimum, maximum, or a saddle point by looking at its eigenvalues.
